```python
# Load all five raw CSV files into PySpark DataFrames

base_path = "/Volumes/workspace/default/t_series_forecasting_data/raw_data"

df_errors = spark.read.option("header", True).option("inferSchema", True).csv(f"{base_path}/PdM_errors.csv")

df_failures = spark.read.option("header", True).option("inferSchema", True).csv(f"{base_path}/PdM_failures.csv")

df_machines = spark.read.option("header", True).option("inferSchema", True).csv(f"{base_path}/PdM_machines.csv")

df_maint = spark.read.option("header", True).option("inferSchema", True).csv(f"{base_path}/PdM_maint.csv")

df_telemetry = spark.read.option("header", True).option("inferSchema", True).csv(f"{base_path}/PdM_telemetry.csv")
```

```python
# Check the columns and data types of all five raw tables

print("===== PdM_errors =====")
df_errors.printSchema()

print("===== PdM_failures =====")
df_failures.printSchema()

print("===== PdM_machines =====")
df_machines.printSchema()

print("===== PdM_maint =====")
df_maint.printSchema()

print("===== PdM_telemetry =====")
df_telemetry.printSchema()
```

```python
# Store the raw errors data as a Bronze Delta table

errors_bronze_path = "/Volumes/workspace/default/t_series_forecasting_data/bronze_data/PdM_errors"

df_errors.write.format("delta").save(errors_bronze_path)
```

```python
# Display only 20 records from the Bronze errors table

display(spark.read.format("delta").load(errors_bronze_path).limit(20))
```

```python
# Store the raw failures data as a Bronze Delta table

failures_bronze_path = "/Volumes/workspace/default/t_series_forecasting_data/bronze_data/PdM_failures"

df_failures.write.format("delta").save(failures_bronze_path)
```

```python
# Display only 20 records from the Bronze failures table

display(spark.read.format("delta").load(failures_bronze_path).limit(20))
```

```python
# Store the raw machine information as a Bronze Delta table

machines_bronze_path = "/Volumes/workspace/default/t_series_forecasting_data/bronze_data/PdM_machines"

df_machines.write.format("delta").save(machines_bronze_path)
```

```python
# Display only 20 records from the Bronze machines table

display(spark.read.format("delta").load(machines_bronze_path).limit(20))
```

```python
# Store the raw maintenance data as a Bronze Delta table

maint_bronze_path = "/Volumes/workspace/default/t_series_forecasting_data/bronze_data/PdM_maint"

df_maint.write.format("delta").save(maint_bronze_path)
```

```python
# Display only 20 records from the Bronze maintenance table

display(spark.read.format("delta").load(maint_bronze_path).limit(20))
```

```python
# Store the raw telemetry data as a Bronze Delta table

telemetry_bronze_path = "/Volumes/workspace/default/t_series_forecasting_data/bronze_data/PdM_telemetry"

df_telemetry.write.format("delta").save(telemetry_bronze_path)
```

```python
# Display only 20 records from the Bronze telemetry table

display(spark.read.format("delta").load(telemetry_bronze_path).limit(20))
```

```python
# Compare raw CSV row counts with Bronze Delta row counts

tables = [
    ("PdM_errors", df_errors, errors_bronze_path),
    ("PdM_failures", df_failures, failures_bronze_path),
    ("PdM_machines", df_machines, machines_bronze_path),
    ("PdM_maint", df_maint, maint_bronze_path),
    ("PdM_telemetry", df_telemetry, telemetry_bronze_path)
]

for table_name, raw_df, bronze_path in tables:
    raw_count = raw_df.count()
    bronze_count = spark.read.format("delta").load(bronze_path).count()
    
    status = "PASS" if raw_count == bronze_count else "CHECK"
    
    print(f"{table_name}: Raw = {raw_count}, Bronze = {bronze_count}, Status = {status}")
```
